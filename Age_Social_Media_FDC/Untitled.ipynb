## What We Found Before Cleaning: Pew vs. ANES

Even though both surveys look almost completely clean at first glance (showing `0%` missing data on almost every column), looking closer at the actual numbers shows that **both datasets hide missing answers behind number codes and measure things on different scales**. 

Before we can combine or compare them, here is what we need to fix:

* **Hidden Missing Answers:** 
  Instead of leaving blank cells (`NaN`), Pew uses **`99`** whenever someone refused to answer across age, gender, education, and social media (on top of 176 people who skipped the social media section completely). Meanwhile, ANES uses **`-7`** for refusals and **`-1`** when a question was skipped

* **Age (Brackets vs. Exact Numbers):** 
  Pew groups people into **4 broad age brackets** (`1` to `4`), while ANES records their **exact age in years** (from `18` up to `80+`). To compare them fairly, we'll need to group the ANES ages into Pew's 4 brackets.

* **Gender (3 Options vs. 2 Options):** 
  Pew includes a **third gender category** (`3`, with 45 respondents) alongside `1` and `2`, whereas ANES only records two (`1` and `2`). We'll also double-check the codebooks to make sure `1` and `2` mean the same gender in both surveys.

* **Education (3 Levels vs. 5 Levels):** 
  Pew splits education into **3 levels**, while ANES uses **5 more detailed levels**. We will need to merge some of the ANES categories so both surveys share the same 3-tier scale (and make sure they are ordered in the same direction).

* **Metro vs. Rural Coding:** 
  Both surveys cleanly track whether someone lives in a metropolitan area without any missing data, but Pew codes it as **`1` vs. `2`** while ANES uses **`1` vs. `0`**.

**Next Step:** Clean up all the `99`, `-1`, and `-7` codes so missing data is handled properly, and put every variable on the exact same scale across both datasets.
